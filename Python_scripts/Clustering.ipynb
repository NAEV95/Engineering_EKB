```python
from Bio.PDB import PDBParser, Selection

def radius_of_gyration(pdb_file):
    # Parse the PDB file
    parser = PDBParser()
    structure = parser.get_structure("structure", pdb_file)

    # Extract the coordinates of all atoms
    atoms = Selection.unfold_entities(structure, 'A')
    coords = [atom.get_coord() for atom in atoms]

    # Calculate the center of mass
    total_mass = sum(atom.mass for atom in atoms)
    center_of_mass = sum(atom.mass * atom.coord for atom in atoms) / total_mass

    # Calculate the radius of gyration
    rg_squared = sum((coord - center_of_mass)**2 for coord in coords) / len(coords)
    rg = rg_squared**0.5

    return rg
```
